Inspect pre-processed GeoTIFFs (bands, scale, grid)

```python
import rasterio
from rasterio.windows import Window

# Positional band order - used ONLY when band names cannot be recovered at all.
# Index = band order in the exported file. Verify/adjust if the run prints a fallback warning.
BAND_ORDER = {
    'intensity': ['HH', 'HV', 'VV', 'VH'],
    'halpha':    ['H', 'A', 'alpha'],         # Entropy, Anisotropy, Alpha
    'yama':      ['Pd', 'Pv', 'Ps', 'Pc'],    # dbl_r, vol_g, surf_b, hlx
}

DECOMP_BANDS = {
    'halpha': {'H': 'Entropy', 'alpha': 'Alpha', 'A': 'Anisotropy'},
    'yama':   {'Pd': 'Yamaguchi_dbl_r', 'Pv': 'Yamaguchi_vol_g',
               'Ps': 'Yamaguchi_surf_b', 'Pc': 'Yamaguchi_hlx'},
}

POL_ORDER = ['HH', 'HV', 'VV', 'VH']

# ---------------------------------------------------------------------------
# Band-name recovery - several channels before falling back to BAND_ORDER.
# SNAP's native GeoTIFF writer stores the full DIMAP header (with band names)
# in the private TIFF tag 65000, which GDAL/rasterio ignore. Channel 4 + 5 read
# that tag / raw bytes directly.
# ---------------------------------------------------------------------------
def _names_from_descriptions(src):
    d = [x for x in src.descriptions]
    return d if (d and any(x for x in d)) else None

def _names_from_tags(src):
    names = []
    for i in range(1, src.count + 1):
        t = src.tags(i) or {}
        nm = None
        for k in ('DESCRIPTION', 'BandName', 'band_name', 'LABEL', 'name'):
            if t.get(k):
                nm = t[k]
                break
        names.append(nm)
    return names if any(names) else None

def _names_from_gdal(path):
    try:
        from osgeo import gdal
    except Exception:
        return None
    try:
        ds = gdal.Open(path)
        if ds is None:
            return None
        names = []
        for i in range(1, ds.RasterCount + 1):
            b = ds.GetRasterBand(i)
            nm = b.GetDescription()
            if not nm:
                md = b.GetMetadata()
                if md:
                    nm = md.get('DESCRIPTION')
            names.append(nm)
        ds = None
        return names if any(names) else None
    except Exception:
        return None

def _read_tiff_tag_bytes(path, target_tag):
    # Minimal TIFF IFD reader (Classic + BigTIFF) to fetch a tag's raw value.
    import struct as _s
    try:
        with open(path, 'rb') as f:
            hdr = f.read(16)
            bo = '<' if hdr[:2] == b'II' else ('>' if hdr[:2] == b'MM' else None)
            if bo is None:
                return None
            magic = _s.unpack(bo + 'H', hdr[2:4])[0]
            if magic == 42:                    # Classic TIFF
                ifd_off = _s.unpack(bo + 'I', hdr[4:8])[0]
                f.seek(ifd_off)
                (ncount,) = _s.unpack(bo + 'H', f.read(2))
                es = 12
            elif magic == 43:                  # BigTIFF
                ifd_off = _s.unpack(bo + 'Q', hdr[8:16])[0]
                f.seek(ifd_off)
                (ncount,) = _s.unpack(bo + 'Q', f.read(8))
                es = 20
            else:
                return None
            endian = 'little' if bo == '<' else 'big'
            for _ in range(ncount):
                ent = f.read(es)
                if len(ent) < es:
                    break
                if magic == 42:
                    tag = _s.unpack(bo + 'H', ent[0:2])[0]
                    typ = _s.unpack(bo + 'H', ent[2:4])[0]
                    cnt = _s.unpack(bo + 'I', ent[4:8])[0]
                    valoff = _s.unpack(bo + 'I', ent[8:12])[0]
                    max_inline = 4
                else:
                    tag = _s.unpack(bo + 'H', ent[0:2])[0]
                    typ = _s.unpack(bo + 'H', ent[2:4])[0]
                    cnt = _s.unpack(bo + 'Q', ent[4:12])[0]
                    valoff = _s.unpack(bo + 'Q', ent[12:20])[0]
                    max_inline = 8
                if tag != target_tag:
                    continue
                if typ in (1, 2, 7):           # BYTE / ASCII / UNDEFINED
                    size = cnt
                else:
                    return None
                if size <= max_inline:
                    return (valoff.to_bytes(max_inline, endian))[:size]
                f.seek(valoff)
                return f.read(size)
    except Exception:
        return None
    return None

def _names_from_tiff65000(path):
    # SNAP private tag 65000 = full DIMAP header XML; extract band names+order.
    raw = _read_tiff_tag_bytes(path, 65000)
    if not raw:
        return None
    try:
        txt = raw.decode('utf-8', 'replace')
    except Exception:
        return None
    import re as _re
    pat = _re.compile(r'BAND_INDEX[ ]*>([0-9]+)<[ ]*[/]BAND_INDEX.*?BAND_NAME[ ]*>([^<]+)<[ ]*[/]BAND_NAME', _re.DOTALL)
    ms = list(pat.finditer(txt))
    if ms:
        pairs = []
        for m in ms:
            try:
                pairs.append((int(m.group(1)), m.group(2).strip()))
            except Exception:
                pass
        if pairs:
            pairs.sort(key=lambda t: t[0])
            return [n for _, n in pairs]
    names = []
    for m in _re.compile(r'BAND_NAME[ ]*>([^<]+)<[ ]*[/]BAND_NAME').finditer(txt):
        n = m.group(1).strip()
        if n:
            names.append(n)
    return names or None

def _names_from_bytes(path):
    # scan the raw file for DIMAP band-name XML (same info as tag 65000)
    import re as _re
    size = os.path.getsize(path)
    if size <= 128 * 1024 * 1024:
        with open(path, 'rb') as f:
            raw = f.read()
        parts = [raw]
    else:
        with open(path, 'rb') as f:
            head = f.read(12 * 1024 * 1024)
            f.seek(max(0, size - 12 * 1024 * 1024))
            tail = f.read()
        parts = [head, tail]
    for part in parts:
        ns = [m.group(1).decode('utf-8', 'replace')
              for m in _re.compile(rb'<BAND_INDEX[ ]*>[0-9]+<[ ]*[/]BAND_INDEX.*?<BAND_NAME[ ]*>([^<]+)<[ ]*[/]BAND_NAME', _re.DOTALL).finditer(part)]
        if ns and all(x for x in ns):
            return ns
    return None

def _names(src, path):
    for fn in (_names_from_descriptions, _names_from_tags, _names_from_gdal,
               _names_from_tiff65000, _names_from_bytes):
        try:
            n = fn(src) if fn in (_names_from_descriptions, _names_from_tags) else fn(path)
            if n:
                return n
        except Exception:
            continue
    return None

def extract_pol(name):
    u = name.upper()
    for pol in POL_ORDER:
        if pol in u:
            return pol
    return None

def resolve_int(src, path):
    desc = _names(src, path)
    if desc and any(extract_pol(d) for d in desc):
        pos = {}
        for i, d in enumerate(desc):
            p = extract_pol(d)
            if p and p not in pos:
                pos[p] = i
        pols = [p for p in POL_ORDER if p in pos]
        return pols, [pos[p] for p in pols], False
    order = list(BAND_ORDER['intensity'])
    return order, list(range(len(order))), True

def resolve_decomp(src, path, kind):
    desc = _names(src, path)
    if desc:
        low = [(d or '').lower() for d in desc]
        out = {}
        for feat, bn in DECOMP_BANDS[kind].items():
            bl = bn.lower()
            idx = [i for i, d in enumerate(low) if d == bl or bl in d]
            if not idx:
                raise ValueError("band '%s' not found in %s" % (bn, low))
            out[feat] = idx[0]
        return out, False
    return {feat: i for i, feat in enumerate(BAND_ORDER[kind])}, True

# ---------------------------------------------------------------------------
# Inspect the three GeoTIFFs
# ---------------------------------------------------------------------------
_GEOTIFFS = {'intensity': INTENSITY_TIF, 'halpha': HALPHA_TIF, 'yama': YAMA_TIF}
_ref = {}
_names_found = {}
print("=== Pre-processed GeoTIFF inspection ===")
for _k, _p in _GEOTIFFS.items():
    with rasterio.open(_p) as _src:
        _desc = _names(_src, _p)
        _names_found[_k] = bool(_desc)
        _ref[_k] = (_src.crs, _src.transform, _src.width, _src.height)
        print()
        print("--- " + _k + " ---")
        print(f"  CRS={_src.crs} size={_src.width}x{_src.height} "
              f"res={_src.res} nodata={_src.nodata} bands={_src.count}")
        print("  bands: " + str(_desc if _desc else [f'Band{i+1}' for i in range(_src.count)]))
        if not _desc:
            print("  [WARN] band names not found anywhere in the file -> positional BAND_ORDER")
        # always print per-band value ranges (diagnostic for content/CRS/grid issues)
        _w = Window(max(0, (_src.width - 300)//2), max(0, (_src.height - 300)//2),
                    min(300, _src.width), min(300, _src.height))
        for _bi in range(_src.count):
            _arr = _src.read(_bi + 1, window=_w).astype(float)
            _v = _arr[np.isfinite(_arr)]
            _v = _v[_v != 0]
            if _v.size:
                _r0 = float(_v.min()); _r1 = float(_v.max())
            else:
                _r0 = float('nan'); _r1 = float('nan')
            print(f"    band {_bi + 1}: sample min={_r0:.4f} max={_r1:.4f}")

for _k in ('halpha', 'yama'):
    print(f"  grid matches intensity: {_ref['intensity'] == _ref[_k]}")

# --- resolve band indices (names if found, else positional) ---
_r_src = rasterio.open(INTENSITY_TIF)
available_pols, INT_BAND_IDX, INT_POSITIONAL = resolve_int(_r_src, INTENSITY_TIF)
sample = _r_src.read(1, window=Window(max(0, (_r_src.width - 500)//2), max(0, (_r_src.height - 500)//2),
                                      min(500, _r_src.width), min(500, _r_src.height))).astype(float)
_r_src.close()
valid = sample[np.isfinite(sample) & (sample != 0)]

DECOMP_IDX = {}
DECOMP_POSITIONAL = {}
for _k in ('halpha', 'yama'):
    with rasterio.open(_GEOTIFFS[_k]) as _src:
        _d, _ispos = resolve_decomp(_src, _GEOTIFFS[_k], _k)
    DECOMP_IDX[_k] = _d
    DECOMP_POSITIONAL[_k] = _ispos

print()
print("Polarizations found: " + str(available_pols) + ("   [POSITIONAL - verify!]" if INT_POSITIONAL else ""))
print("INT_BAND_IDX: " + str(INT_BAND_IDX))
print("DECOMP_IDX: " + str(DECOMP_IDX))
print("names_found_per_file: " + str(_names_found))
if any(DECOMP_POSITIONAL.values()):
    print("[WARN] decomposition band order is POSITIONAL (BAND_ORDER) - check the per-band stats above.")

if INT_POSITIONAL:
    with rasterio.open(INTENSITY_TIF) as _t:
        _w0 = Window(max(0, (_t.width - 300)//2), max(0, (_t.height - 300)//2), 300, 300)
        _arr = _t.read(window=_w0).astype(float)
    _flat = _arr.reshape(_arr.shape[0], -1)
    _m = np.isfinite(_flat).all(axis=0)
    _cm = np.corrcoef(_flat[:, _m])
    print("  intensity band correlation matrix (rows/cols = band 1..N):")
    print("    " + "".join(["%8d" % (i + 1) for i in range(_cm.shape[0])]))
    for _i in range(_cm.shape[0]):
        print("  %d " % (_i + 1) + "".join(["%8.3f" % _cm[_i, _j] for _j in range(_cm.shape[1])]))

if valid.size > 0:
    vmin, vmax, vmean = float(valid.min()), float(valid.max()), float(valid.mean())
    has_negative = vmin < 0
    is_db = bool(has_negative)
    print(f"Intensity sample: min={vmin:.4f} max={vmax:.4f} mean={vmean:.4f} "
          f"has_negative={has_negative} -> {'dB' if is_db else 'linear'}")
else:
    is_db = True
    print("WARNING: empty sample region; assuming dB scale")
```

## 2-2. Sample Backscatter + Decomposition at Tree Coordinates ; Build Dataset


For each tree `(id, Long, Lat)` in the label CSVs:
- Extract a **3x3-pixel window** (`EXTRACT_WINDOW`) and take the **mean dB** within it (ignoring nodata)
- Extract the same window's **mean/std** of the polarimetric decomposition bands:
  `H, alpha, A` (Cloude-Pottier) and `Ps, Pd, Pv, Pc` (Yamaguchi 4-component)
- Derive mechanism ratios (`Pv/Ptotal`, `Pd/Pv`, `Ps/Pv`, `H*(1-A)`)
- Retain the existing intensity features: window stats, GLCM texture, local-background anomaly, ratios, RVI


```python
import rasterio
from rasterio.warp import transform as _warp
from rasterio.windows import Window

print("Sampling intensity (dB) + polarimetric decomposition features at tree centroids.")
print(f"  Intensity GeoTIFF: {os.path.basename(INTENSITY_TIF)}")
print(f"  H/Alpha GeoTIFF  : {os.path.basename(HALPHA_TIF)}")
print(f"  Yamaguchi GeoTIFF: {os.path.basename(YAMA_TIF)}")
print(f"  Pols: {available_pols} | scale: {'dB' if is_db else 'linear'}")
print(f"  Extract window: {EXTRACT_WINDOW}x{EXTRACT_WINDOW} (mean/std) | "
      f"texture {TEXTURE_WINDOW}x{TEXTURE_WINDOW} | background {BACKGROUND_WINDOW}x{BACKGROUND_WINDOW}")

_src_int = rasterio.open(INTENSITY_TIF)
_src_ha  = rasterio.open(HALPHA_TIF)
_src_ya  = rasterio.open(YAMA_TIF)

half     = EXTRACT_WINDOW // 2
tex_half = TEXTURE_WINDOW // 2
bg_half  = BACKGROUND_WINDOW // 2
n_pol    = len(available_pols)

GLCM_FEATS = ['contrast', 'dissimilarity', 'homogeneity', 'energy', 'correlation']

def _glcm_features(window_2d):
    try:
        from skimage.feature import graycomatrix, graycoprops
    except ImportError:
        return None
    vmin, vmax = np.nanmin(window_2d), np.nanmax(window_2d)
    if not np.isfinite(vmin) or not np.isfinite(vmax) or vmax - vmin < 1e-6:
        return None
    mask = np.isfinite(window_2d)
    if mask.sum() < 4:
        return None
    q = np.zeros(window_2d.shape, dtype=np.uint8)
    q[mask] = np.round((window_2d[mask] - vmin) / (vmax - vmin) * 31).astype(np.uint8)
    glcm = graycomatrix(q, distances=GLCM_DISTANCES, angles=GLCM_ANGLES,
                        levels=32, symmetric=True, normed=True)
    feats = {}
    for fname in GLCM_FEATS:
        vals = graycoprops(glcm, fname)
        feats[fname] = float(np.nanmean(vals))
    return feats

def _win_stats(src, row, col, win, idx):
    w = Window(max(0, col - win // 2), max(0, row - win // 2), win, win)
    w = w.intersection(Window(0, 0, src.width, src.height))
    if w.width == 0 or w.height == 0:
        return np.nan, np.nan
    data = src.read(window=w, indexes=idx + 1).astype(float).squeeze()
    if src.nodata is not None:
        data[data == src.nodata] = np.nan
    data[data <= -99] = np.nan
    valid = data[np.isfinite(data)]
    if valid.size == 0:
        return np.nan, np.nan
    return float(valid.mean()), float(valid.std(ddof=0))

datasets = {}
for location in LOCATIONS:
    csv_path = f"{LABEL_DIR}/{location} Classification.csv"
    df = pd.read_csv(csv_path)
    print()
    print(f"--- {location} ({len(df)} trees) ---")

    _lon = df['Long'].values.tolist()
    _lat = df['Lat'].values.tolist()
    xs_s, ys_s = _warp("EPSG:4326", _src_int.crs, _lon, _lat)    # intensity CRS
    xs_ha, ys_ha = _warp("EPSG:4326", _src_ha.crs, _lon, _lat)   # halpha CRS
    xs_ya, ys_ya = _warp("EPSG:4326", _src_ya.crs, _lon, _lat)   # yamaguchi CRS
    _DEC_COORDS = {'halpha': (xs_ha, ys_ha), 'yama': (xs_ya, ys_ya)}

    stat_names = ['mean', 'std', 'min', 'max', 'p25', 'p50', 'p75']
    samples = {s: np.full((len(df), n_pol), np.nan, dtype=float) for s in stat_names}
    glcm_store = {f: np.full((len(df), n_pol), np.nan, dtype=float) for f in GLCM_FEATS}
    bg_median = np.full((len(df), n_pol), np.nan, dtype=float)

    # ---- intensity dB sampling (existing feature family) ----
    for idx, (x, y) in enumerate(zip(xs_s, ys_s)):
        row, col = _src_int.index(x, y)

        w = Window(max(0, col - half), max(0, row - half), EXTRACT_WINDOW, EXTRACT_WINDOW)
        w = w.intersection(Window(0, 0, _src_int.width, _src_int.height))
        if w.width == 0 or w.height == 0:
            continue
        data = _src_int.read(window=w).astype(float)
        if _src_int.nodata is not None:
            data[data == _src_int.nodata] = np.nan
        data[data <= -99] = np.nan
        for _j in range(n_pol):
            _bb = INT_BAND_IDX[_j]
            valid = data[_bb][np.isfinite(data[_bb])]
            if valid.size > 0:
                samples['mean'][idx, _j] = valid.mean()

        wt = Window(max(0, col - tex_half), max(0, row - tex_half),
                    TEXTURE_WINDOW, TEXTURE_WINDOW)
        wt = wt.intersection(Window(0, 0, _src_int.width, _src_int.height))
        if wt.width > 0 and wt.height > 0:
            data_t = _src_int.read(window=wt).astype(float)
            if _src_int.nodata is not None:
                data_t[data_t == _src_int.nodata] = np.nan
            data_t[data_t <= -99] = np.nan
            for _j in range(n_pol):
                _bb = INT_BAND_IDX[_j]
                win = data_t[_bb]
                valid = win[np.isfinite(win)]
                if valid.size == 0:
                    continue
                samples['std'][idx, _j] = valid.std()
                samples['min'][idx, _j] = valid.min()
                samples['max'][idx, _j] = valid.max()
                samples['p25'][idx, _j] = np.nanpercentile(win, 25)
                samples['p50'][idx, _j] = np.nanpercentile(win, 50)
                samples['p75'][idx, _j] = np.nanpercentile(win, 75)
            for _j in range(n_pol):
                _bb = INT_BAND_IDX[_j]
                g = _glcm_features(data_t[_bb])
                if g:
                    for fname in GLCM_FEATS:
                        glcm_store[fname][idx, _j] = g[fname]

        # local-background annulus (for the anomaly / detrend feature)
        wbg = Window(max(0, col - bg_half), max(0, row - bg_half),
                     BACKGROUND_WINDOW, BACKGROUND_WINDOW)
        wbg = wbg.intersection(Window(0, 0, _src_int.width, _src_int.height))
        if wbg.width > 0 and wbg.height > 0:
            d_bg = _src_int.read(window=wbg).astype(float)
            if _src_int.nodata is not None:
                d_bg[d_bg == _src_int.nodata] = np.nan
            d_bg[d_bg <= -99] = np.nan
            amask = np.ones((wbg.height, wbg.width), dtype=bool)
            r0 = max(0, row - half - wbg.row_off)
            c0 = max(0, col - half - wbg.col_off)
            r1 = min(wbg.height, row + half - wbg.row_off + 1)
            c1 = min(wbg.width,  col + half - wbg.col_off + 1)
            if r1 > r0 and c1 > c0:
                amask[r0:r1, c0:c1] = False
            for _j in range(n_pol):
                _bb = INT_BAND_IDX[_j]
                vals = d_bg[_bb][amask]
                vals = vals[np.isfinite(vals)]
                if vals.size > 0:
                    bg_median[idx, _j] = np.median(vals)

    # ---- assign intensity columns ----
    for i, pol in enumerate(available_pols):
        df[pol] = samples['mean'][:, i]
        if USE_WINDOW_STATS:
            for s in ['std', 'min', 'max', 'p25', 'p50', 'p75']:
                df[f"{pol}_{s}"] = samples[s][:, i]
        if USE_GLCM:
            for fname in GLCM_FEATS:
                df[f"{pol}_{fname}"] = glcm_store[fname][:, i]
        if USE_ANOMALY:
            df[f"{pol}_anom"] = samples['mean'][:, i] - bg_median[:, i]

    # ---- polarimetric decomposition sampling (mean/std of EXTRACT_WINDOW) ----
    dec_cols = []
    for _k, _s in [('halpha', _src_ha), ('yama', _src_ya)]:
        _cx, _cy = _DEC_COORDS[_k]
        for feat in DECOMP_IDX[_k]:
            idx = DECOMP_IDX[_k][feat]
            mu = np.full(len(df), np.nan)
            sd = np.full(len(df), np.nan)
            for i in range(len(df)):
                row, col = _s.index(_cx[i], _cy[i])
                mu[i], sd[i] = _win_stats(_s, row, col, EXTRACT_WINDOW, idx)
            df[f"{feat}_mean"] = mu
            df[f"{feat}_std"]  = sd
            dec_cols += [f"{feat}_mean", f"{feat}_std"]

    # diagnostic: flag decomp features that came back entirely NaN
    for _c in dec_cols:
        if _c in df.columns and df[_c].isna().all():
            print(f"  [WARN] decomp feature '{_c}' is ALL NaN - sample window missed valid pixels")
            print("         -> check that the decomposition GeoTIFFs' CRS/grid matches the intensity GeoTIFF")

    # mechanism ratios (from window means, zero/NaN guarded)
    Ps = df['Ps_mean']; Pd = df['Pd_mean']
    Pv = df['Pv_mean']; Pc = df['Pc_mean']
    tot = (Ps + Pd + Pv + Pc).replace(0, np.nan)
    df['Pv_ratio'] = Pv / tot
    df['Pd_ratio'] = Pd / tot
    df['Ps_ratio'] = Ps / tot
    df['Pc_ratio'] = Pc / tot
    with np.errstate(divide='ignore', invalid='ignore'):
        df['Pd_over_Pv'] = Pd / Pv.replace(0, np.nan)
        df['Ps_over_Pv'] = Ps / Pv.replace(0, np.nan)
    df['H_times_1mA'] = df['H_mean'] * (1 - df['A_mean'])
    dec_cols += ['Pv_ratio', 'Pd_ratio', 'Ps_ratio', 'Pc_ratio',
                 'Pd_over_Pv', 'Ps_over_Pv', 'H_times_1mA']

    # ---- intensity ratios / indices ----
    if 'HH' in df.columns and 'HV' in df.columns:
        df['HHHVRatio'] = df['HH'] - df['HV']
    if 'HH' in df.columns and 'VV' in df.columns:
        df['HHVVRatio'] = df['HH'] - df['VV']
    if 'HV' in df.columns and 'VV' in df.columns:
        df['HVVVRatio'] = df['HV'] - df['VV']
    if 'VH' in df.columns and 'HV' in df.columns:
        df['VHHVRatio'] = df['VH'] - df['HV']
    if USE_RVI and all(p in df.columns for p in ['HH', 'HV', 'VV']):
        hh_lin = 10 ** (df['HH'] / 10.0)
        hv_lin = 10 ** (df['HV'] / 10.0)
        vv_lin = 10 ** (df['VV'] / 10.0)
        denom = hh_lin + vv_lin + 2 * hv_lin
        df['RVI'] = np.where(denom > 0, 8 * hv_lin / denom, np.nan)
    else:
        df['RVI'] = np.nan

    df['Health'] = df['Class']
    df['Tier'] = 0

    # ---- feature column list (intensity + decomposition) ----
    feature_cols = []
    for pol in available_pols:
        feature_cols.append(pol)
        if USE_ANOMALY:
            feature_cols.append(f"{pol}_anom")
        if USE_WINDOW_STATS:
            for s in ['std', 'min', 'max', 'p25', 'p50', 'p75']:
                feature_cols.append(f"{pol}_{s}")
        if USE_GLCM:
            for fname in GLCM_FEATS:
                feature_cols.append(f"{pol}_{fname}")
    feature_cols += ['HHHVRatio', 'HHVVRatio', 'HVVVRatio', 'VHHVRatio']
    if USE_RVI:
        feature_cols.append('RVI')
    feature_cols += dec_cols

    final_cols = ['id', 'Long', 'Lat'] + [c for c in feature_cols if c in df.columns] + ['Health', 'Tier']
    final_df = df[final_cols].copy()

    # per-scene z-score of numeric feature columns (label-free normalization)
    if USE_ZSCORE:
        for c in feature_cols:
            if c not in final_df.columns:
                continue
            col = final_df[c].astype(float)
            mu2 = col.mean(); sd2 = col.std(ddof=0)
            if sd2 and np.isfinite(sd2) and sd2 > 1e-9:
                final_df[c] = (col - mu2) / sd2
            else:
                final_df[c] = 0.0

    datasets[location] = final_df

    valid = final_df['HH'].notna()
    n_valid = int(valid.sum()); n_nan = int((~valid).sum())
    print(f"  Valid dB: {n_valid}/{len(final_df)}   NaN: {n_nan}")
    print(f"  Class distribution: {final_df['Health'].value_counts().to_dict()}")
    print(f"  Features ({len(final_cols)} total = intensity + {len(dec_cols)} decomposition)")

_src_int.close(); _src_ha.close(); _src_ya.close()

# Write per-location datasets (window + tag in filename auto cache-invalidates)
for location, df in datasets.items():
    output_path = f"{OUTPUT_DIR}/dataset_{location}_w{EXTRACT_WINDOW}_{DATASET_TAG}.csv"
    df.to_csv(output_path, index=False)
    print(f"Wrote {output_path} ({len(df)} rows)")
```

## 3-3. Shared Evaluation Harness
Per-location metrics, threshold tuning, holdout & CV evaluation.


```python
# ---- Metrics ------------------------------------------------------------------
METRIC_NAMES = [
    'accuracy','balanced_accuracy',
    'precision_unhealthy','recall_unhealthy','f1_unhealthy',
    'f1_macro',
    'roc_auc','pr_auc',
    'brier_score','cohen_kappa',
    'runtime_train_s','runtime_pred_s',
    'tn','fp','fn','tp','threshold_used',
]

def compute_metrics(y_true, y_pred, y_proba, train_time=0.0, pred_time=0.0, threshold=0.5):
    cm = confusion_matrix(y_true, y_pred)
    if cm.shape == (2,2):
        tn, fp, fn, tp = cm.ravel()
    else:
        tn = fp = fn = tp = 0

    if y_proba.ndim == 1:
        pos_proba = y_proba
    else:
        pos_proba = y_proba[:, 1]

    return {
        'accuracy':            accuracy_score(y_true, y_pred),
        'balanced_accuracy':   balanced_accuracy_score(y_true, y_pred),
        'precision_unhealthy': precision_score(y_true, y_pred, zero_division=0),
        'recall_unhealthy':    recall_score(y_true, y_pred, zero_division=0),
        'f1_unhealthy':        f1_score(y_true, y_pred, zero_division=0),
        'f1_macro':            f1_score(y_true, y_pred, average='macro', zero_division=0),
        'roc_auc':             roc_auc_score(y_true, pos_proba),
        'pr_auc':              average_precision_score(y_true, pos_proba),
        'brier_score':         brier_score_loss(y_true, pos_proba),
        'cohen_kappa':         cohen_kappa_score(y_true, y_pred),
        'runtime_train_s':     train_time,
        'runtime_pred_s':      pred_time,
        'tn': int(tn), 'fp': int(fp), 'fn': int(fn), 'tp': int(tp),
        'threshold_used':      threshold,
    }

# ---- Threshold tuning ----------------------------------------------------------
def find_best_threshold(y_true, y_proba):
    """Pick threshold on PR curve that maximises F1 for minority class."""
    if y_proba.ndim == 1:
        pos_proba = y_proba
    else:
        pos_proba = y_proba[:, 1]
    prec_list, rec_list, thresholds = precision_recall_curve(y_true, pos_proba)
    n = len(thresholds)
    prec = prec_list[:n]
    rec  = rec_list[:n]
    with np.errstate(divide='ignore', invalid='ignore'):
        f1s = 2 * (prec * rec) / (prec + rec + 1e-12)
    best_idx = np.nanargmax(f1s)
    return thresholds[best_idx], f1s[best_idx]

def predict_with_threshold(model, X, threshold):
    y_proba = model.predict_proba(X)
    if y_proba.ndim == 1:
        return (y_proba >= threshold).astype(int), y_proba
    return (y_proba[:, 1] >= threshold).astype(int), y_proba

# ---- Pipeline builder (no SMOTE) ------------------------------------------------
def build_pipeline(model, use_scaling=False):
    steps = []
    if use_scaling:
        steps.append(('scaler', StandardScaler()))
    steps.append(('model', model))
    return Pipeline(steps)

# ---- Per-location Holdout evaluation ---------------------------------------------
def eval_holdout_loc(pipeline, X_tr, y_tr, X_te, y_te):
    t0 = time.time()
    pipeline.fit(X_tr, y_tr)
    train_time = time.time() - t0

    threshold = 0.5
    if USE_THRESHOLD_TUNING:
        y_tr_proba = pipeline.predict_proba(X_tr)
        threshold, _ = find_best_threshold(y_tr, y_tr_proba)

    t0 = time.time()
    y_pred, y_proba = predict_with_threshold(pipeline, X_te, threshold)
    pred_time = time.time() - t0

    return compute_metrics(y_te, y_pred, y_proba, train_time, pred_time, threshold)

# ---- Per-location CV evaluation -------------------------------------------------
def eval_cv_loc(pipeline_factory, loc, fold_start=0, fold_end=None, groups=None):
    X = X_LOC[loc]
    y = y_LOC[loc]
    results = []
    if groups is not None:
        gkf, blocks = groups
        folds = list(gkf.split(X, y, groups=blocks))
    else:
        folds = list(cvs[loc].split(X, y))
    if fold_end is None:
        fold_end = len(folds)
    for fold_idx in range(fold_start, fold_end):
        train_idx, test_idx = folds[fold_idx]
        X_tr, X_te = X.iloc[train_idx], X.iloc[test_idx]
        y_tr, y_te = y.iloc[train_idx], y.iloc[test_idx]

        pipe = pipeline_factory()
        t0 = time.time()
        pipe.fit(X_tr, y_tr)
        train_time = time.time() - t0

        threshold = 0.5
        if USE_THRESHOLD_TUNING:
            y_tr_proba = pipe.predict_proba(X_tr)
            threshold, _ = find_best_threshold(y_tr, y_tr_proba)

        t0 = time.time()
        y_pred, y_proba = predict_with_threshold(pipe, X_te, threshold)
        pred_time = time.time() - t0

        m = compute_metrics(y_te, y_pred, y_proba, train_time, pred_time, threshold)
        m['fold'] = fold_idx
        results.append(m)
    return pd.DataFrame(results)

def eval_spatial_cv_loc(pipeline_factory, loc):
    """Run per-block spatial CV -- the HONEST within-location estimate.

    Builds folds on the fly (via make_spatial_folds) so this works even after a
    kernel restart, and tags results with eval_type='spatial_cv'.
    """
    if not USE_SPATIAL_CV:
        return None
    global spatial_cvs
    if 'spatial_cvs' not in globals() or not spatial_cvs or loc not in spatial_cvs:
        if 'spatial_cvs' not in globals():
            spatial_cvs = {}
        spatial_cvs[loc] = make_spatial_folds(loc)
    df = eval_cv_loc(pipeline_factory, loc, groups=spatial_cvs[loc])
    if df is not None and not df.empty:
        df = df.copy()
        df['eval_type'] = 'spatial_cv'
    return df

# ---- Save / summary helpers -----------------------------------------------------
def save_results(result_df, model_name, eval_type, location='combined'):
    d = os.path.join(RESULTS_DIR, location)
    os.makedirs(d, exist_ok=True)
    path = f"{d}/{model_name}_{eval_type}.csv"
    if isinstance(result_df, dict):
        result_df = pd.DataFrame([result_df])
    result_df.to_csv(path, index=False)
    print(f"  Saved: {path}")
    return path

def cv_summary_df(cv_df):
    score_cols = [c for c in METRIC_NAMES if c in cv_df.columns]
    m = cv_df[score_cols].mean()
    s = cv_df[score_cols].std()
    return {f"{col}_mean": m[col] for col in score_cols} | \
           {f"{col}_std" : s[col] for col in score_cols}

class_weight_fn = lambda neg, pos: {0: 1.0, 1: MINORITY_WEIGHT_MULT * (neg / max(pos, 1))}

print("Harness ready.")
```

```python
# ---- Plot helpers ------------------------------------------------------------------
def plot_confusion_matrix(cm, title, ax, labels=('Healthy (0)','Unhealthy (1)')):
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=labels, yticklabels=labels)
    ax.set_title(title)
    ax.set_ylabel('True'); ax.set_xlabel('Predicted')

def plot_roc_curve(fpr, tpr, auc_val, label, ax):
    ax.plot(fpr, tpr, label=f'{label} (AUC={auc_val:.3f})', linewidth=2)
    ax.plot([0,1],[0,1],'k--',alpha=0.3)
    ax.set_xlabel('FPR'); ax.set_ylabel('TPR'); ax.set_title('ROC Curves')
    ax.legend(loc='lower right',fontsize=8)
    ax.set_xlim([-0.02,1.02]); ax.set_ylim([-0.02,1.02])

def plot_pr_curve(prec, rec, ap_val, label, baseline, ax):
    ax.plot(rec, prec, label=f'{label} (AP={ap_val:.3f})', linewidth=2)
    ax.axhline(y=baseline,color='gray',linestyle='--',alpha=0.3,label=f'No skill ({baseline:.3f})')
    ax.set_xlabel('Recall'); ax.set_ylabel('Precision'); ax.set_title('PR Curves')
    ax.legend(loc='lower left',fontsize=8)
    ax.set_xlim([-0.02,1.02]); ax.set_ylim([-0.02,1.02])

def plot_roc_pr_for_model(pipeline, X_tr, y_tr, X_te, y_te, config_label, ax_roc, ax_pr):
    pipeline.fit(X_tr, y_tr)
    y_proba = pipeline.predict_proba(X_te)
    pos_proba = y_proba[:,1] if y_proba.ndim > 1 else y_proba
    fpr,tpr,_ = roc_curve(y_te, pos_proba)
    roc_auc = roc_auc_score(y_te, pos_proba)
    prec,rec,_ = precision_recall_curve(y_te, pos_proba)
    pr_auc = average_precision_score(y_te, pos_proba)
    baseline_val = sum(y_te==1)/len(y_te)
    plot_roc_curve(fpr,tpr,roc_auc,config_label,ax_roc)
    plot_pr_curve(prec,rec,pr_auc,config_label,baseline_val,ax_pr)

print("Plot helpers ready.")
```

# 4. Machine Learning
Each model is trained **per location**, evaluated on holdout + within-location CV.

## ML1 -- Random Forest


```python
# ============================================================================
# ML1 -- Random Forest  (revised to fix bias toward the Healthy majority)
# ============================================================================
# Diagnosis of the previous result
#   * The data is ~10:1 Healthy:Unhealthy, so a plain RF can get high accuracy
#     by predicting "Healthy" for everything - this is exactly the observed
#     bias, and accuracy/f1_macro hide it.  The metrics that matter here are
#     PR-AUC, Recall(Unhealthy) and F1(Unhealthy).
#   * A lot of engineered features (window stats, GLCM, ratios, RVI) are highly
#     correlated.  Feeding all of them at once adds noise and dilutes the few
#     informative ones, which pushes the model even harder toward the majority.
#   * Out-of-scene trees can leave missing dB -> NaN features, which
#     RandomForestClassifier cannot handle.
#
# Fixes applied below
#   1. Class weighting offsets the imbalance (weights from per-location counts).
#   2. Redundant / constant / all-NaN columns are pruned INSIDE the pipeline
#      (learned per-CV-fold on the training part only), so no held-out leakage.
#   3. A median SimpleImputer fills the remaining NaN before the model.
#   4. Wider max_features (+0.5) & deeper tree regularisation (min_samples_leaf)
#      so strong discriminators get used and overfit to the majority is reduced.
#   5. Tuned with the same impute+dedup pipeline inside CV (honest scores).
# ============================================================================
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, mutual_info_classif
from sklearn.base import BaseEstimator, TransformerMixin

# --- drop constant & near-collinear feature columns (X only, no label leak) ---
class DropRedundantFeatures(BaseEstimator, TransformerMixin):
    """Removes constant / all-NaN / highly-correlated columns.

    Only uses X (never y), so it introduces no label leakage.  For each group of
    correlated features it keeps the highest-variance member, which reduces the
    noise injected by the large set of SAR/GLCM features.
    """
    def __init__(self, var_thr=1e-6, corr_thr=0.90):
        self.var_thr = var_thr
        self.corr_thr = corr_thr
        self.kept_cols_ = None

    def fit(self, X, y=None):
        Xf = X.astype(float)
        var = Xf.var(axis=0).fillna(-1.0)          # all-NaN columns -> -1 -> dropped
        keep = var > self.var_thr
        cols = list(Xf.columns[keep])
        Xc = Xf[cols]
        if len(cols) <= 1:
            self.kept_cols_ = cols
            return self
        corr = Xc.corr().abs()                     # pairwise Pearson (ignores NaN)
        dropped = set()
        order = Xc.var().sort_values(ascending=False).index.tolist()
        for i, col in enumerate(order[:-1]):       # keep highest-variance member
            if col in dropped:
                continue
            for col2 in order[i + 1:]:
                if col2 in dropped:
                    continue
                if corr.loc[col, col2] >= self.corr_thr:
                    dropped.add(col2)
        self.kept_cols_ = [c for c in order if c not in dropped]
        return self

    def transform(self, X):
        return X[self.kept_cols_]

# --- supervised MI top-k filter (POC-4 validated) --------------------------------
# Under the HONEST spatial-block CV used by the Section-4 restart harness:
#   * a supervised mutual-information top-k stage beats BOTH the raw feature dump
#     AND the dedup-only pipeline on PR-AUC / recall / balanced-accuracy.
#   * PCA was the WORST reducer under spatial CV -> deliberately NOT used here.
#   * k=20 was the sweet spot (k~3 over-pruned). The score is computed on the
#     training fold only (this pipeline is fit/X_.fit inside CV), so it never
#     sees held-out data -> no label/data leakage.
SELECT_K = 20

def _mi_score(X, y, _rs=RANDOM_STATE):
    return mutual_info_classif(X, y, random_state=_rs)

# RF pipeline: dedup -> impute -> supervised MI top-k -> RandomForest
# (all steps kept in every split so dedup / imputation / selection never leak)
def make_rf_pipe(rf, n_select=SELECT_K):
    return Pipeline([
        ('dedup',  DropRedundantFeatures(corr_thr=0.90)),
        ('impute', SimpleImputer(strategy='median')),
        ('select', SelectKBest(_mi_score, k=n_select)),
        ('model',  rf),
    ])

RF_PARAM_GRID = {
    'model__n_estimators':       [200, 500],
    'model__max_depth':          [None, 10, 20],
    'model__min_samples_split':  [2, 5],
    'model__min_samples_leaf':   [1, 2],
    'model__max_features':       ['sqrt', 0.5],   # allow more discriminators/split
}

rf_best_params = {}
rf_all_holdout = []
rf_all_cv = []

for loc in LOCATIONS:
    print(f"\n{'='*70}\n  LOCATION: {loc}\n{'='*70}")
    Xtr, ytr = SPLITS[loc]['X_train'], SPLITS[loc]['y_train']
    Xte, yte = SPLITS[loc]['X_test'],  SPLITS[loc]['y_test']
    neg, pos = int(sum(ytr == 0)), int(sum(ytr == 1))
    # class weights computed on the TRAIN split only (no holdout usage)
    cw = {0: 1.0, 1: MINORITY_WEIGHT_MULT * (neg / max(pos, 1))}
    print(f"  Train: Healthy={neg}, Unhealthy={pos} "
          f"(ratio {neg / max(pos, 1):.1f}:1) -> class_weight[1]={cw[1]:.2f}")

    # Grid search over the FULL pipeline so dedup + imputation run inside CV
    # (held-out folds never feed feature selection / imputation).
    rf_gs = GridSearchCV(
        make_rf_pipe(RandomForestClassifier(class_weight=cw,
                                            random_state=RANDOM_STATE, n_jobs=-1)),
        RF_PARAM_GRID, cv=3, scoring='average_precision', n_jobs=-1,
    )
    rf_gs.fit(Xtr, ytr)
    rf_best_params[loc] = {k.replace('model__', ''): v
                           for k, v in rf_gs.best_params_.items()}
    n_kept = len(rf_gs.best_estimator_.named_steps['dedup'].kept_cols_)
    n_sel  = rf_gs.best_estimator_.named_steps['select'].k
    print(f"  Features: {len(feature_names[loc])} -> {n_kept} after dedup -> {n_sel} after MI select")
    print(f"  Best params: {rf_best_params[loc]}")
    print(f"  Best CV avg_precision: {rf_gs.best_score_:.4f}")

    def make_rf_default(loc=loc):
        neg, pos = sum(y_LOC[loc] == 0), max(sum(y_LOC[loc] == 1), 1)
        return RandomForestClassifier(
            n_estimators=500,
            class_weight={0: 1.0, 1: MINORITY_WEIGHT_MULT * (neg / pos)},
            random_state=RANDOM_STATE, n_jobs=-1,
        )

    def make_rf_tuned(loc=loc):
        neg, pos = sum(y_LOC[loc] == 0), max(sum(y_LOC[loc] == 1), 1)
        return RandomForestClassifier(
            class_weight={0: 1.0, 1: MINORITY_WEIGHT_MULT * (neg / pos)},
            random_state=RANDOM_STATE, n_jobs=-1,
            **rf_best_params[loc],
        )

    configs = [('default', make_rf_default), ('tuned', make_rf_tuned)]
    for cfg_name, factory in configs:
        print(f"\n--- RF {loc} {cfg_name} ---")
        pipe = make_rf_pipe(factory())
        m = eval_holdout_loc(pipe, Xtr, ytr, Xte, yte)
        m['config'] = cfg_name; m['model'] = 'rf'; m['location'] = loc
        rf_all_holdout.append(m)
        print(f"  Holdout: PR-AUC={m['pr_auc']:.4f}, F1(Unhealthy)={m['f1_unhealthy']:.4f}, "
              f"Recall={m['recall_unhealthy']:.4f}, Precision={m['precision_unhealthy']:.4f}, "
              f"thresh={m['threshold_used']:.3f}")

        def _make(f=factory): return make_rf_pipe(f())
        cv_df = eval_cv_loc(_make, loc)
        cv_df['config'] = cfg_name; cv_df['model'] = 'rf'; cv_df['location'] = loc
        rf_all_cv.append(cv_df)
        print(f"  CV:      PR-AUC={cv_df['pr_auc'].mean():.4f} +/- {cv_df['pr_auc'].std():.4f}, "
              f"F1(Unhealthy)={cv_df['f1_unhealthy'].mean():.4f} +/- {cv_df['f1_unhealthy'].std():.4f}")

        # Spatial block CV (honest within-location estimate)
        sp_cv = eval_spatial_cv_loc(_make, loc)
        if sp_cv is not None:
            sp_cv['config'] = cfg_name; sp_cv['model'] = 'rf'; sp_cv['location'] = loc
            sp_cv['eval_type'] = 'spatial_cv'
            rf_all_cv.append(sp_cv)
            print(f"  SpatialCV: PR-AUC={sp_cv['pr_auc'].mean():.4f} +/- {sp_cv['pr_auc'].std():.4f}, "
                  f"F1(Unhealthy)={sp_cv['f1_unhealthy'].mean():.4f} +/- {sp_cv['f1_unhealthy'].std():.4f}")

# Save
rf_holdout_df = pd.DataFrame(rf_all_holdout)
rf_cv_df = pd.concat(rf_all_cv, ignore_index=True)
for loc in LOCATIONS:
    save_results(rf_holdout_df[rf_holdout_df['location'] == loc], 'rf', 'holdout', loc)
    save_results(rf_cv_df[rf_cv_df['location'] == loc], 'rf', 'cv', loc)
print("\n=== RF Complete ===")
```

```python
print(f"XGBoost version: {xgb.__version__}")
if USE_FOCAL_LOSS:
    print(f"Using focal loss (gamma={FOCAL_GAMMA})")
else:
    print("Using binary:logistic + scale_pos_weight")
```

```python
# ---- Focal loss custom objective -------------------------------------
def _focal_loss_obj(y_pred, dtrain):
    """Correct binary focal loss gradient & hessian for XGBoost.

    Focal loss:  FL = -alpha_t * (1 - p_t)^gamma * log(p_t)
    where p_t = p if y==1 else (1-p), alpha_t = alpha if y==1 else (1-alpha).
    Gradient (w.r.t. logit z):  dFL/dz = alpha_t * (1-p_t)^gamma * (gamma * p_t * log(p_t) + p_t - 1)
    Hessian:                     d2FL/dz2 = alpha_t * (1-p_t)^(gamma-1) * (gamma * p_t * log(p_t) + p_t - 1) * (gamma * log(p_t) + 1) * p_t * (1-p_t)
    """
    if hasattr(dtrain, 'get_label'):
        # Native XGBoost API: (preds, dtrain) — dtrain is DMatrix
        y_true = dtrain.get_label()
    else:
        # Sklearn wrapper XGBClassifier 2.x+: (y_true, y_pred) — numpy arrays
        y_true = y_pred
        y_pred = dtrain
    p = 1.0 / (1.0 + np.exp(-y_pred))  # sigmoid
    p = np.clip(p, 1e-7, 1 - 1e-7)
    gamma = FOCAL_GAMMA
    alpha = 0.25  # focal-loss alpha (class balance term)

    p_t = np.where(y_true == 1, p, 1 - p)
    alpha_t = np.where(y_true == 1, alpha, 1 - alpha)
    log_pt = np.log(p_t)

    # gradient
    grad = alpha_t * (1 - p_t) ** gamma * (gamma * p_t * log_pt + p_t - 1)
    # hessian (second derivative)
    hess = alpha_t * (1 - p_t) ** (gamma - 1) * (gamma * p_t * log_pt + p_t - 1) * (gamma * log_pt + 1) * p_t * (1 - p_t)
    hess = np.clip(hess, 1e-6, None)  # keep positive for XGBoost

    # apply minority weight multiplier
    pos = y_true == 1
    if MINORITY_WEIGHT_MULT != 1.0 and pos.any():
        grad[pos] *= MINORITY_WEIGHT_MULT
        hess[pos] *= MINORITY_WEIGHT_MULT
    return grad, hess

def _focal_eval(y_pred, dtrain):
    """Eval metric (logloss-style) for focal loss — required when using custom obj."""
    if hasattr(dtrain, 'get_label'):
        # Native XGBoost API: (preds, dtrain) — dtrain is DMatrix
        y_true = dtrain.get_label()
    else:
        # Sklearn wrapper XGBClassifier 2.x+: (y_true, y_pred) — numpy arrays
        y_true = y_pred
        y_pred = dtrain
    p = 1.0 / (1.0 + np.exp(-y_pred))
    p = np.clip(p, 1e-16, 1-1e-16)
    return 'logloss_eval', -np.mean(y_true * np.log(p) + (1-y_true) * np.log(1-p))

# ---- Per-location XGBoost -------------------------------------------------
XGB_PARAM_GRID = {
    'n_estimators': [300],
    'max_depth': [3, 6, 9],
    'learning_rate': [0.05, 0.1, 0.2],
    'subsample': [1.0],
    'colsample_bytree': [0.8, 1.0],
}

xgb_best_params = {}
xgb_all_holdout = []
xgb_all_cv = []

for loc in LOCATIONS:
    print(f"\n{'='*70}\n  LOCATION: {loc}\n{'='*70}")
    Xtr, ytr = SPLITS[loc]['X_train'], SPLITS[loc]['y_train']
    Xte, yte = SPLITS[loc]['X_test'], SPLITS[loc]['y_test']

    scale_weight = sum(y_LOC[loc]==0) / max(sum(y_LOC[loc]==1),1) * MINORITY_WEIGHT_MULT
    print(f"scale_pos_weight: {scale_weight:.2f}")

    # XGBoost base
    if USE_FOCAL_LOSS:
        xgb_base = XGBClassifier(
            device='cuda', tree_method='hist',
            objective=_focal_loss_obj, eval_metric=_focal_eval,
            random_state=RANDOM_STATE, n_jobs=1
        )
    else:
        xgb_base = XGBClassifier(
            device='cuda', tree_method='hist',
            scale_pos_weight=scale_weight, eval_metric='aucpr',
            random_state=RANDOM_STATE, n_jobs=1
        )

    # GridSearch
    xgb_gs = GridSearchCV(
        xgb_base, XGB_PARAM_GRID, cv=3, scoring='average_precision', n_jobs=1
    )
    xgb_gs.fit(Xtr, ytr)
    xgb_best_params[loc] = xgb_gs.best_params_
    print(f"Best params: {xgb_gs.best_params_}")
    print(f"Best CV f1_macro: {xgb_gs.best_score_:.4f}")

    def make_xgb_default(loc=loc):
        sw = sum(y_LOC[loc]==0) / max(sum(y_LOC[loc]==1),1) * MINORITY_WEIGHT_MULT
        if USE_FOCAL_LOSS:
            return XGBClassifier(
                n_estimators=500, max_depth=6, learning_rate=0.1,
                device='cuda', tree_method='hist',
                objective=_focal_loss_obj, eval_metric=_focal_eval,
                random_state=RANDOM_STATE, n_jobs=1
            )
        else:
            return XGBClassifier(
                n_estimators=500, max_depth=6, learning_rate=0.1,
                device='cuda', tree_method='hist',
                scale_pos_weight=sw, eval_metric='aucpr',
                random_state=RANDOM_STATE, n_jobs=1
            )
    def make_xgb_tuned(loc=loc):
        sw = sum(y_LOC[loc]==0) / max(sum(y_LOC[loc]==1),1) * MINORITY_WEIGHT_MULT
        if USE_FOCAL_LOSS:
            return XGBClassifier(
                device='cuda', tree_method='hist',
                objective=_focal_loss_obj, eval_metric=_focal_eval,
                random_state=RANDOM_STATE, n_jobs=1,
                **xgb_best_params[loc]
            )
        else:
            return XGBClassifier(
                device='cuda', tree_method='hist',
                scale_pos_weight=sw, eval_metric='aucpr',
                random_state=RANDOM_STATE, n_jobs=1,
                **xgb_best_params[loc]
            )

    configs = [('default', make_xgb_default), ('tuned', make_xgb_tuned)]
    for cfg_name, factory in configs:
        print(f"\n--- XGB {loc} {cfg_name} ---")
        pipe = build_pipeline(factory(), use_scaling=False)
        m = eval_holdout_loc(pipe, Xtr, ytr, Xte, yte)
        m['config'] = cfg_name; m['model'] = 'xgb'; m['location'] = loc
        xgb_all_holdout.append(m)
        print(f"  Holdout: PR-AUC={m['pr_auc']:.4f}, F1(Unhealthy)={m['f1_unhealthy']:.4f}, "
              f"Recall={m['recall_unhealthy']:.4f}, thresh={m['threshold_used']:.3f}")

        def _make(f=factory): return build_pipeline(f(), use_scaling=False)
        cv_df = eval_cv_loc(_make, loc)
        cv_df['config'] = cfg_name; cv_df['model'] = 'xgb'; cv_df['location'] = loc
        xgb_all_cv.append(cv_df)
        print(f"  CV:      PR-AUC={cv_df['pr_auc'].mean():.4f} ± {cv_df['pr_auc'].std():.4f}, "
              f"F1(Unhealthy)={cv_df['f1_unhealthy'].mean():.4f} ± {cv_df['f1_unhealthy'].std():.4f}")

        # Spatial block CV (honest within-location estimate)
        sp_cv = eval_spatial_cv_loc(_make, loc)
        if sp_cv is not None:
            sp_cv['config'] = cfg_name; sp_cv['model'] = 'xgb'; sp_cv['location'] = loc
            sp_cv['eval_type'] = 'spatial_cv'
            xgb_all_cv.append(sp_cv)
            print(f"  SpatialCV: PR-AUC={sp_cv['pr_auc'].mean():.4f} ± {sp_cv['pr_auc'].std():.4f}, "
                  f"F1(Unhealthy)={sp_cv['f1_unhealthy'].mean():.4f} ± {sp_cv['f1_unhealthy'].std():.4f}")

# Save
xgb_holdout_df = pd.DataFrame(xgb_all_holdout)
xgb_cv_df = pd.concat(xgb_all_cv, ignore_index=True)
for loc in LOCATIONS:
    save_results(xgb_holdout_df[xgb_holdout_df['location']==loc], 'xgb', 'holdout', loc)
    save_results(xgb_cv_df[xgb_cv_df['location']==loc], 'xgb', 'cv', loc)
print("\n=== XGBoost Complete ===")
```
